# Aula 5: Conjuntos numéricos — números racionais [🎥](https://www.youtube.com/watch?v=N9V-rbeWAgE&list=PL8aWdrmfXHiKJeft-RcH9OXXBBgoGdQy7&index=21)

## Introdução

Na Aula 4, vimos que a divisão de dois inteiros pode produzir um resultado que não é inteiro. Por exemplo, a equação $4x=1$ tem solução $x=\frac14$ no conjunto dos números racionais. Nesta aula, estudaremos esse conjunto, suas propriedades e a relação entre frações e representações decimais.

## 5.1 Conjunto dos números racionais

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Conjunto dos números racionais.</strong> Um número é racional quando pode ser escrito como o quociente de dois inteiros, com denominador diferente de zero.
</div>

Indicamos o conjunto dos números racionais por $\mathbb{Q}$.

$$
\mathbb{Q}=\left\{\frac{a}{b}\;\middle|\;a,b\in\mathbb{Z}\ \text{e}\ b\ne 0\right\}.
$$

O numerador pode ser zero, como em $\frac05=0$. O denominador deve ser diferente de zero.

#### Exemplo 1

Os números abaixo pertencem a $\mathbb{Q}$, pois podem ser escritos como frações de inteiros.

$$
2=\frac21,\qquad
-3=\frac{-3}{1},\qquad
0{,}25=\frac{25}{100}=\frac14,\qquad
0{,}333\ldots=\frac13.
$$

Na última igualdade, o algarismo $3$ se repete indefinidamente. Veremos como obter essa fração na seção 5.4.

## 5.2 Subconjuntos dos racionais

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Subconjuntos de racionais.</strong> Assim como na Aula 4, usamos os sinais de mais e de menos para indicar o lado do zero considerado e o asterisco para excluir o zero.
</div>

- **Racionais não nulos**: $\mathbb{Q}^{*}=\mathbb{Q}-\{0\}$.
- **Racionais não negativos**: $\mathbb{Q}_{+}=\{x\in\mathbb{Q}\mid x\geq 0\}$.
- **Racionais não positivos**: $\mathbb{Q}_{-}=\{x\in\mathbb{Q}\mid x\leq 0\}$.
- **Racionais positivos**: $\mathbb{Q}_{+}^{*}=\{x\in\mathbb{Q}\mid x>0\}$.
- **Racionais negativos**: $\mathbb{Q}_{-}^{*}=\{x\in\mathbb{Q}\mid x<0\}$.

Por exemplo, $-\frac34$ pertence a $\mathbb{Q}_{-}^{*}$ e $0$ pertence a $\mathbb{Q}_{+}$ e a $\mathbb{Q}_{-}$.

## 5.3 Propriedades dos números racionais

### Inclusão

Todo natural é inteiro, e todo inteiro $n$ é racional, pois $n=\frac{n}{1}$. Assim,

$$\mathbb{N}\subsetneq\mathbb{Z}\subsetneq\mathbb{Q}.$$

As duas inclusões são próprias. Por exemplo, $-3$ é inteiro e não é natural, e $\frac14$ é racional e não é inteiro. Em diagramas de Venn:<br><br>

### Fechamento

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Fechamento.</strong> A soma, a diferença e o produto de dois racionais são racionais. O quociente também é racional quando o divisor é diferente de zero.
</div>

Sejam $\frac{a}{b}$ e $\frac{c}{d}$ dois racionais, com $a,b,c,d\in\mathbb{Z}$, $b\ne 0$ e $d\ne 0$. Então,

$$
\begin{aligned}
\frac{a}{b}+\frac{c}{d}&=\frac{ad+bc}{bd},\\[5pt]
\frac{a}{b}-\frac{c}{d}&=\frac{ad-bc}{bd},\\[5pt]
\frac{a}{b}\cdot\frac{c}{d}&=\frac{ac}{bd},\\[5pt]
\frac{a}{b}\div\frac{c}{d}&=\frac{a}{b}\cdot\frac{d}{c}=\frac{ad}{bc}\qquad(c\ne 0).
\end{aligned}
$$

Os numeradores e denominadores dos resultados são inteiros. Como $b$ e $d$ são diferentes de zero, $bd\ne 0$. Na divisão, também precisamos de $c\ne 0$, pois $\frac cd$ é o divisor e $c$ passa a ser o denominador da fração invertida.

#### Exemplo 2

Para somar $3$ e $\frac{2}{5}$, escrevemos $3=\frac{3}{1}$ e usamos o denominador comum $5$.

$$
\begin{aligned}
3 + \frac{2}{5}
& = \frac{3}{1} + \frac{2}{5}\\
& = \frac{3\cdot5 + 2\cdot1}{5}\\
& = \frac{17}{5}
\end{aligned}
$$

Podemos considerar o quociente de $0,25$ por $0,15$ usando os racionais:

$$
\begin{aligned}
0,25 \div 0,15
& = \large{\frac{\frac{25}{100}}{\frac{15}{100}}}\\
& = \frac{25}{100}\cdot\frac{100}{15}\\
& = \frac{25}{15}\\
& = \frac{5}{3}
\end{aligned}
$$

### Um racional entre dois racionais

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Densidade dos racionais.</strong> Entre dois racionais distintos sempre existe outro número racional.
</div>

Por exemplo, podemos exibir um racional entre $0,7$ e $0,8$. Note que, $0,71$ satisfaz essa condição. De fato,

$$0,70<0,71<0,80 \Leftrightarrow \frac{70}{100} < \frac{71}{100} < \frac{80}{100}.$$

De modo mais geral, podemos obter um racional entre dois números da seguinte forma:

Se $r,s\in\mathbb{Q}$ e $r<s$, podemos tomar a média aritmética $m=\frac{r+s}{2}$. Pelas propriedades de fechamento, $m$ é racional. Somando $r$ a $r<s$, obtemos $2r<r+s$. Somando $s$ à mesma desigualdade, obtemos $r+s<2s$. Logo,

$$
2r<r+s<2s
\qquad\Rightarrow\qquad
r<\frac{r+s}{2}<s.
$$

Por exemplo, entre os racionais $2$ e $3$, temos o racional:

$$2 < \frac{2+3}{2} < 3 \Leftrightarrow 2 < \frac{5}{2} < 3.$$

### Oposto e inverso multiplicativo

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Oposto e inverso.</strong> Somar um racional ao seu oposto dá zero. Multiplicar um racional não nulo pelo seu inverso multiplicativo dá um.
</div>

O **oposto** de $r\in\mathbb{Q}$ é $-r$, pois $r+(-r)=0$. Em particular, o oposto de zero é o próprio zero.

Por exemplo, o posto de $\frac{2}{3}$ é $-\frac{2}{3}$. E o oposto de $-\frac{1}{2}$ é $\frac{1}{2}$. Em todos dois casos,

i) $\frac{2}{3} + \left(-\frac{2}{3}\right) = \frac{2}{3} - \frac{2}{3} = \frac{2-2}{3} = 0$;

ii) $-\frac{1}{2} + \frac{1}{2} = \frac{-1+1}{2} = 0$.

Se $r\ne 0$, seu **inverso multiplicativo** é $r^{-1}=\frac1r$, pois $r\cdot\frac1r=1$. Assim, se $r=\frac{a}{b}$ com $a\ne 0$ e $b\ne 0$, então $r^{-1}=\frac{b}{a}$. Veja três exemplos.

i) $2\cdot\frac{1}{2}=1$;

ii) $\frac{1}{4}\cdot4=1$;

iii) $\left(-\frac{3}{4}\right)\cdot\left(-\frac{4}{3}\right)=1$.

O zero não tem inverso multiplicativo, pois $0\cdot s=0$ para todo $s\in\mathbb{Q}$.

## 5.4 Dízimas periódicas e frações geratrizes

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Dízima periódica.</strong> É uma representação decimal infinita em que, a partir de certa casa, um bloco de algarismos se repete indefinidamente. O menor bloco que se repete é o período.
</div>

A barra destaca o período. Uma dízima é **simples** quando o período começa logo depois da vírgula. É **composta** quando há algarismos entre a vírgula e o período. Esse trecho é chamado de **parte não periódica** ou **anteperíodo**.

- $0,333\ldots = 0,\overline{3} \rightarrow$ Período: $3$, Parte não periódica: nenhuma
- $0,121212\ldots = 0,\overline{12} \rightarrow$ Período: $12$, Parte não períódica: nenhuma
- $0,41323131\ldots =0,4\overline{13} \rightarrow$ Período: $13$, Parte não periódica: $13$
- $0,42153153153\ldots = 0,42\overline{153} \rightarrow$ Período: $153$, Parte não periódica: $42$

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Fração geratriz.</strong> É uma fração que representa a dízima. Para encontrá-la, deslocamos a vírgula com potências de dez até obter duas expressões com a mesma parte decimal. Ao subtraí-las, eliminamos a parte que se repete.
</div>

#### Exemplo 3

Determine as frações geratrizes de $0{,}\overline{3}$ e de $0{,}\overline{12}$.

##### Solução

Como $0{,}\overline{3}$ tem período de um algarismo, multiplicamos $x$ por $10$ e subtraímos $x$.

$$
\begin{aligned}
x&=0{,}333\ldots,\\
10x&=3{,}333\ldots,\\
10x-x&=3,\\
x&=\frac39=\frac13.
\end{aligned}
$$

Para $0{,}\overline{12}$, o período tem dois algarismos. Por isso usamos $100$.

$$
\begin{aligned}
y&=0{,}121212\ldots,\\
100y&=12{,}121212\ldots,\\
100y-y&=12,\\
y&=\frac{12}{99}=\frac{4}{33}.
\end{aligned}
$$

#### Exemplo 4

Determine as frações geratrizes de $0{,}4\overline{13}$ e de $0{,}42\overline{153}$.

##### Solução

Para $x=0{,}4\overline{13}$, multiplicamos por $10$ para passar pelo algarismo $4$. A partir daí, deslocamos a vírgula mais duas casas, pois o período $13$ tem dois algarismos.

$$
\begin{aligned}
10x&=4{,}131313\ldots,\\
1000x&=413{,}131313\ldots,\\
1000x-10x&=413-4,\\
990x&=409,\\
x&=\frac{409}{990}.
\end{aligned}
$$

Para $y=0{,}42\overline{153}$, multiplicamos primeiro por $100$ para passar pelos algarismos $42$. Para deslocar a vírgula mais três casas, usamos $100\,000$ em relação ao número original.

$$
\begin{aligned}
100y&=42{,}153153\ldots,\\
100\,000y&=42\,153{,}153153\ldots,\\
100\,000y-100y&=42\,153-42,\\
99\,900y&=42\,111,\\
y&=\frac{42\,111}{99\,900}=\frac{4679}{11\,100}.
\end{aligned}
$$

#### Exemplo 5

Escreva $0{,}47\overline{32}$ como fração.

##### Solução

Podemos fazer a mesma subtração diretamente. No numerador, subtraímos $47$ de $4732$. No denominador, escrevemos um $9$ para cada algarismo do período e um $0$ para cada algarismo do anteperíodo. Neste caso, $9900=10^4-10^2$.

$$
0{,}47\overline{32}=\frac{4732-47}{9900}
=\frac{4685}{9900}=\frac{937}{1980}.
$$

#### Exemplo 6

Calcule $0{,}25\cdot\frac23-3\cdot0{,}\overline{3}$.

##### Solução

Escrevemos $0{,}25=\frac14$ e $0{,}\overline{3}=\frac13$. Efetuamos as multiplicações antes da subtração.

$$
\begin{aligned}
0{,}25\cdot\frac23-3\cdot0{,}\overline{3}
&=\frac14\cdot\frac23-3\cdot\frac13\\[4pt]
&=\frac16-1\\[4pt]
&=-\frac56.
\end{aligned}
$$

## Exercícios Resolvidos

#### Exercício 1 [🎥](https://www.youtube.com/watch?v=lPMaIKuqVAs&list=PL8aWdrmfXHiKJeft-RcH9OXXBBgoGdQy7&index=23)

Determine o valor da expressão.

$$
\frac{2{,}333\ldots}{1{,}1666\ldots}-4\cdot1{,}333\ldots
$$

##### Solução

Primeiro, obtemos as frações geratrizes das três dízimas.

**(I)** Para $x=2{,}\overline{3}$, temos $10x=23{,}\overline{3}$. Subtraindo as duas igualdades, obtemos $9x=23-2=21$. Logo, $x=\frac{21}{9}=\frac73$.

**(II)** Para $y=1{,}1\overline{6}$, temos $10y=11{,}\overline{6}$ e $100y=116{,}\overline{6}$. Assim, $90y=116-11=105$ e $y=\frac{105}{90}=\frac76$.

**(III)** Para $z=1{,}\overline{3}$, temos $10z=13{,}\overline{3}$. Logo, $9z=13-1=12$ e $z=\frac{12}{9}=\frac43$.

Agora substituímos as dízimas por suas frações e calculamos.

$$
\begin{aligned}
\frac{2{,}333\ldots}{1{,}1666\ldots}-4\cdot1{,}333\ldots
&=\frac73\div\frac76-4\cdot\frac43\\[4pt]
&=\frac73\cdot\frac67-\frac{16}{3}\\[4pt]
&=2-\frac{16}{3}\\[4pt]
&=-\frac{10}{3}.
\end{aligned}
$$

#### Exercício 2 [🎥](https://www.youtube.com/watch?v=9azkZtYoBgU&list=PL8aWdrmfXHiKJeft-RcH9OXXBBgoGdQy7&index=24)

Se $p$ e $q$ são inteiros e $q\ne 0$, qual das afirmações é sempre verdadeira? Considere cada quociente onde ele está definido.

**a)** $\dfrac pq$ é inteiro.

**b)** $\dfrac{p}{p+q}$ é inteiro.

**c)** $\dfrac{p+q}{q}$ é inteiro.

**d)** $\dfrac pq$ é inteiro se, e somente se, existe um inteiro $k$ tal que $p=kq$.

**e)** Se $\dfrac pq$ é inteiro, então $\dfrac qp$ é inteiro.

##### Solução

**a)** Falsa. Para $p=2$ e $q=3$, temos $\frac pq=\frac23\notin\mathbb{Z}$.

**b)** Falsa. Para os mesmos valores, $\frac{p}{p+q}=\frac25\notin\mathbb{Z}$.

**c)** Falsa. Para $p=2$ e $q=3$, temos $\frac{p+q}{q}=\frac53\notin\mathbb{Z}$.

**d)** Verdadeira. Se $\frac pq$ é inteiro, podemos chamar esse inteiro de $k$. Como $q\ne 0$, multiplicamos $\frac pq=k$ por $q$ e obtemos $p=kq$. Reciprocamente, se $p=kq$ para algum $k\in\mathbb{Z}$, dividimos a igualdade por $q$ e obtemos $\frac pq=k\in\mathbb{Z}$.

**e)** Falsa. Para $p=4$ e $q=2$, temos $\frac pq=\frac42=2\in\mathbb{Z}$, mas $\frac qp=\frac24=\frac12\notin\mathbb{Z}$.

Portanto, a resposta é a alternativa **d)**.

#### Exercício 3 [🎥](https://www.youtube.com/watch?v=hO6zJk2v9jg&list=PL8aWdrmfXHiKJeft-RcH9OXXBBgoGdQy7&index=25)

Considere o segmento $\overline{AB}$ de medida $u>0$. Descreva um procedimento para obter um ponto $C$ sobre a semirreta $\overrightarrow{AB}$ tal que

$$AC=\frac95u.$$

##### Solução

1. Divida $\overline{AB}$ em cinco partes iguais. Cada parte mede $\frac{u}{5}$.
2. A partir de $A$, marque na semirreta $\overrightarrow{AB}$ nove segmentos consecutivos de comprimento $\frac{u}{5}$. A quinta marca coincide com $B$. A nona marca é o ponto $C$.

Assim, $C$ fica depois de $B$, a quatro partes de $B$, e a distância de $A$ até $C$ é

$$AC=9\cdot\frac{u}{5}=\frac95u.$$
