## Introdução

Nas aulas anteriores, estudamos os números naturais, inteiros e racionais. À medida que ampliamos o conjunto numérico em que trabalhamos, podemos resolver novas equações. Vamos retomar essa ideia e investigar se a equação $x^2-2=0$ tem solução em $\mathbb{Q}$.

## 6.1 Equações e conjuntos numéricos

#### Exemplo 1

Resolva as três equações e identifique, entre $\mathbb{N}$, $\mathbb{Z}$ e $\mathbb{Q}$, o menor conjunto que contém a solução de cada uma.

$$
\text{(a)}\quad x-4=0
\qquad
\text{(b)}\quad x+4=0
\qquad
\text{(c)}\quad 2x+1=0.
$$

##### Solução

**a)** Somamos $4$ aos dois membros e obtemos $x=4$. Como $4\in\mathbb{N}$, a solução é natural.

**b)** Subtraímos $4$ dos dois membros e obtemos $x=-4$. A solução pertence a $\mathbb{Z}$, e a equação em $\mathbb{N}$ tem conjunto solução vazio.

**c)** Subtraímos $1$ dos dois membros e dividimos por $2$.

$$
2x+1=0
\quad\Longrightarrow\quad
2x=-1
\quad\Longrightarrow\quad
x=-\frac{1}{2}.
$$

A solução pertence a $\mathbb{Q}$ e fica fora de $\mathbb{Z}$.

#### A equação $x^2-2=0$

Somando $2$ aos dois membros, chegamos a $x^2=2$. A raiz quadrada $\sqrt{2}$ é o valor positivo cujo quadrado é $2$. O valor $-\sqrt{2}$ também satisfaz a equação.

A pergunta que guia a aula é **$\sqrt{2}$ pode ser escrito como uma fração de inteiros?**

Para responder a essa pergunta, usaremos duas proposições sobre a paridade de um inteiro e de seu quadrado.

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Proposições</strong>
<ol>
  <li>Um número inteiro é par se, e somente se, seu quadrado é par.</li>
  <li>Um número inteiro é ímpar se, e somente se, seu quadrado é ímpar.</li>
</ol>
</div>

##### Demonstração

$\Rightarrow)$ **Se $n$ é par, então $n^2$ é par.** Sendo $n$ par, existe $k\in\mathbb{Z}$ tal que $n=2k$. Logo,

$$
n^2=(2k)^2=4k^2=2(2k^2).
$$

Como $2k^2$ é inteiro, $n^2$ é par.

$\Leftarrow)$ **Se $n^2$ é par, então $n$ é par.** Suponhamos que $n^2$ seja par e, por absurdo, que $n$ seja ímpar. Então $n=2k+1$ para algum $k\in\mathbb{Z}$. Desenvolvendo o quadrado, temos

$$
\begin{aligned}
n^2&=(2k+1)^2\\
   &=4k^2+4k+1\\
   &=2(2k^2+2k)+1.
\end{aligned}
$$

Como $2k^2+2k$ é inteiro, $n^2$ seria ímpar, em contradição com a hipótese de que $n^2$ é par. Logo, $n$ é par. Isso demonstra a Proposição 1.

Deixamos a demonstração da Proposição 2 como exercício para o leitor.

## 6.2 Demonstração da irracionalidade de $\sqrt{2}$

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Demonstração por absurdo.</strong> Para demonstrar uma afirmação, supomos que ela é falsa e examinamos as consequências dessa suposição. Se chegamos a uma contradição, concluímos que a afirmação é verdadeira.
</div>

### A escolha de uma fração irredutível

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Fração irredutível.</strong> Uma fração é irredutível quando o máximo divisor comum do numerador e do denominador é 1.
</div>

#### Exemplo 2

Simplifique a fração $\dfrac{18}{12}$ até obter uma fração irredutível.

##### Solução

Como $\operatorname{mdc}(18,12)=6$, dividimos o numerador e o denominador por $6$.

$$
\frac{18}{12}=\frac{18\div 6}{12\div 6}=\frac{3}{2}.
$$

A fração $\frac{3}{2}$ é irredutível, pois $\operatorname{mdc}(3,2)=1$. Esse procedimento permite representar qualquer número racional por uma fração irredutível.

### Demonstração

Suponhamos, por absurdo, que $\sqrt{2}$ seja racional. Podemos escolher inteiros $p$ e $q$ com $q\ne 0$ e $\operatorname{mdc}(|p|,|q|)=1$ tais que

$$
\sqrt{2}=\frac{p}{q}.
$$

Elevando os dois membros ao quadrado e multiplicando por $q^2$, obtemos

$$
2=\frac{p^2}{q^2}
\Rightarrow
p^2=2q^2.
\tag{1}
$$

Vamos examinar as duas possibilidades para o numerador $p$.

#### Caso 1: $p$ é par

Se $p$ é par, podemos escrever $p=2k$ para algum $k\in\mathbb{Z}$. Substituímos na igualdade (1).

$$
2q^2=p^2=(2k)^2=4k^2
\Rightarrow
q^2=2k^2.
$$

Como $k^2$ é inteiro, $q^2$ é par. Pela Proposição 1, $q$ também é par. Assim, $2$ divide tanto $p$ quanto $q$, o que contradiz $\operatorname{mdc}(|p|,|q|)=1$.

#### Caso 2: $p$ é ímpar

Todo inteiro ímpar tem a forma $2k+1$, com $k\in\mathbb{Z}$. Por exemplo, $19=2\cdot 9+1$ e $5=2\cdot 2+1$.

Escrevendo $p=2k+1$ na igualdade (1), obtemos

$$
2q^2=(2k+1)^2.
$$

O membro esquerdo é par, pois é o dobro de um inteiro. O membro direito é ímpar, pois, pela Proposição 2, o quadrado de um inteiro ímpar também é ímpar. A igualdade é impossível.

#### Conclusão

Todo inteiro é par ou ímpar. Como os dois casos levam a contradições, a suposição de que $\sqrt{2}$ é racional é falsa. Portanto,

$$
\boxed{\sqrt{2}\notin\mathbb{Q}.}
$$

O número $-\sqrt{2}$ também fica fora de $\mathbb{Q}$. Se fosse racional, seu oposto $\sqrt{2}$ seria racional. Assim, o conjunto solução de $x^2-2=0$ em $\mathbb{Q}$ é $\varnothing$.
