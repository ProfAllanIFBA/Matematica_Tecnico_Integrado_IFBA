# Aula 3: Diferença de dois conjuntos e conjunto complementar [🎥](https://www.youtube.com/watch?v=W7GcVXm4mVg)

## Introdução

Na aula anterior, estudamos a união e a intersecção de conjuntos. Agora vamos estudar a diferença de dois conjuntos e usá-la para definir o conjunto complementar. Em seguida, veremos o que acontece quando tomamos o complementar de uma união ou de uma intersecção.

## 3.1 Diferença de dois conjuntos

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Definição.</strong> A diferença de $A$ e $B$, nessa ordem, é o conjunto dos elementos que pertencem a $A$ e não pertencem a $B$. Indicamos essa diferença por $A-B$ e lemos “$A$ menos $B$”.
</div>

$$A-B=\{x\mid x\in A \text{ e } x\notin B\}.$$

Em um diagrama de Venn, $A-B$ corresponde à parte de $A$ que fica fora de $B$. Para obter $B-A$, partimos de $B$ e retiramos os elementos que também estão em $A$.

#### Exemplo 1

Considere os conjuntos $A=\{1,2,3\}$, $B=\{1,6\}$, $C=\{1,2,3,4,5\}$ e $D = \{4,7\}$. Determine:

a) $A-B$

b) $B-A$

c) $C-A$

d) $A-C$

e) $A-D$

##### Solução

a) O único elemento de $A$ que também pertence a $B$ é $1$. Retirando-o de $A$, obtemos $A-B=\{2,3\}$.

b) Retirando de $B$ o elemento $1$, obtemos $B-A=\{6\}$.

c) Os elementos de $A$ que aparecem em $C$ são $1$, $2$ e $3$. Assim, $C-A=\{4,5\}$.

d) Note que todo elemento de $A$ é também elemento de $C$. Neste caso, $A-C = \emptyset$.

e) Note que $A$ e $D$ não tem qualquer elemento em comum. Neste casos, $A-D = A$.

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Propriedades.</strong> Da definição de diferença, obtemos as propriedades a seguir.
</div>

i) Se $A\cap B=\emptyset$, então $A-B=A$ e $B-A=B$. Nesse caso, os conjuntos são disjuntos.

ii) Se $A\subseteq B$, então $A-B=\emptyset$, pois todo elemento de $A$ pertence a $B$.

iii) $A-\emptyset=A$ e $\emptyset-A=\emptyset$.

iv) $A-A=\emptyset$.

## 3.2 Conjunto complementar

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Definição.</strong> Se $A\subseteq B$, o complementar de $A$ em relação a $B$, indicado por $\mathcal{C}_B^A$, é o conjunto dos elementos de $B$ que faltam em $A$.
</div>

$$A \subseteq B \Leftrightarrow \mathcal{C}_B^A=B-A=\{x\in B\mid x\notin A\}.$$

Para calcular uma diferença $B-A$, podemos usar quaisquer conjuntos $A$ e $B$. Para empregar a notação $\mathcal{C}_B^A$, verificamos primeiro se $A\subseteq B$.

#### Exemplo 2

Considere novamente $A=\{1,2,3\}$, $B=\{1,6\}$ e $C=\{1,2,3,4,5\}$. Verifique se existem:

a) $\mathcal{C}_C^A$

b) $\mathcal{C}_B^A$

c) $\mathcal{C}_C^B$

##### Solução

a) Como $A\subseteq C$, existe o complementar $C_C^A=C-A=\{4,5\}$.

b) Como $2\in A$ e $2\notin B$, temos $A\nsubseteq B$. Portanto, $C_B^A$ não existe.

c) Como $6\in B$ e $6\notin C$, temos $B\nsubseteq C$. Portanto, $C_C^B$ não existe.

## 3.3 Complementar em relação ao universo

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Notação.</strong> Fixado um universo $U$, escrevemos $\overline{A}$ para indicar o complementar de $A$ em relação a $U$. Como $A\subseteq U$, esse complementar sempre existe.
</div>

$$\overline{A}=\mathcal{C}_U^A=U-A.$$

A barra sempre se refere ao universo escolhido. Por exemplo, para $A=\{1,2\}$, temos $\overline A=\{3\}$ se $U=\{1,2,3\}$. Se $U=\{1,2,3,4\}$, obtemos $\overline A=\{3,4\}$.

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Propriedades.</strong> O complementar do universo é o vazio, o complementar do vazio é o universo e o complementar do complementar de $A$ é o próprio $A$.
</div>

$$\overline{U}=\emptyset,\qquad \overline{\emptyset}=U,\qquad \overline{\overline{A}}=A.$$

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Leis de De Morgan.</strong> O complementar de uma união é a intersecção dos complementares. O complementar de uma intersecção é a união dos complementares.
</div>

$$\overline{A\cup B}=\overline A\cap\overline B$$

$$\overline{A\cap B}=\overline A\cup\overline B$$

Um elemento fica fora de $A\cup B$ quando fica fora dos dois conjuntos. Já para ficar fora de $A\cap B$, basta ficar fora de pelo menos um deles. Acompanhe a figura a seguir para ver as regiões correspondentes às Leis de De Morgan para dois conjuntos A e B com intersecção não vazia.<br><br>

#### Exemplo 3

Considere $U=\{1,2,3,4\}$, $A=\{1,2\}$ e $B=\{2,3\}$. Verifique a primeira lei de De Morgan nesse caso.

##### Solução

Como $A\cup B=\{1,2,3\}$, temos $\overline{A\cup B}=\{4\}$.

Calculando os complementares separadamente, obtemos $\overline A=\{3,4\}$ e $\overline B=\{2,3\}$. Logo, $\overline A\cap\overline B=\{4\}$, como prevê a primeira lei.

#### Exercício 1 [🎥](https://www.youtube.com/watch?v=99jMtytZ-FA)

Um conjunto universo $U$ tem $n(U)=44$. Três subconjuntos $A$, $B$ e $C$ de $U$ satisfazem as condições a seguir:

$$
\begin{aligned}
n(A)&=18, & n(A\cap B)&=9,\\
n(B)&=22, & n(A\cap C)&=8,\\
n(\overline{A\cup B\cup C})&=9, & n(B\cap C)&=11,\\
n(A\cap B\cap C)&=6.
\end{aligned}
$$

Determine o número de elementos de $C$.

##### Solução

Vamos contar os elementos de cada região, começando pela intersecção dos três conjuntos, que contém $6$ elementos, isto é, em $A \cup B \cup C$. Também sabemos que $9$ elementos ficam fora dos três conjuntos, isto é, em $\overline{A \cup B \cup C}. Esses devem ser os primeiros a serem fixados.

Em cada intersecção de dois conjuntos, precisamos descontar os $6$ elementos comuns aos três. Assim, obtemos:

Em $(A \cap B)-C$ temos $9-6 = 3$ elementos;

Em $(A \cap C)-B$ temos $8-6 = 2$ elementos;

Em $(B \cap C)-A$ temos $11-6 = 5$ elementos;


Dos $18$ elementos de $A$, já contamos $3+2+6$. Logo, $18-(3+2+6)=7$ pertencem somente a $A$.

Dos $22$ elementos de $B$, já contamos $3+5+6$. Logo, $22-(3+5+6)=8$ pertencem somente a $B$.

Temos o seguinte esquema:<br><br>

<center> <img src="fig6.jpg" alt="Logo do IFBA" width="600" height="200"> </center><br>


Se $x$ é o número de elementos que pertencem somente a $C$, a soma de todas as regiões do universo é

$$
\begin{aligned}
7+3+6+2+5+8+x+9=44
&\Rightarrow 40 + x = 44\\
&\Rightarrow x = 44 - 40\\
&\Rightarrow x = 4\\
\end{aligned}
$$

Somando as quatro regiões que compõem $C$, obtemos

$$n(C)=2+6+5+4=17.$$

#### Exercício 2 [🎥](https://www.youtube.com/watch?v=1D219-NsDE8)

Sejam $A$, $B$ e $C$ subconjuntos de $X=\{1,2,3\}$, tais que

Determine $Y=A\cap B\cap C$.

$3 \in C \cap (B - A) \Rightarrow 3 \in C \cap B \text{ e } 3 \notin A \Rightarrow 3 \notin A \cap B \cap C.$

Cada elemento de $X$ fica fora de um dos três conjuntos. Como $A$, $B$ e $C$ estão contidos em $X$, esses são todos os elementos que poderiam pertencer à intersecção. Acompanhe o esquema a seguir:<br><br>

#### Exercício 3 [🎥](https://www.youtube.com/watch?v=roIYeRMTe18)

Considere o diagrama a seguir:<br><br>

<center> <img src="fig8.jpg" alt="Logo do IFBA" width="200" height="200"> </center><br>

A região destacada corresponde a:

a) $(\overline{A \cup B}) \cup C$

b) $\overline{B \cup C}$

c) $\overline{C} \cap \overline{B} \cap \overline{A}$

d) $A - B \cup C$

e) $A - A \cap B \cap C$

##### Solução

Consideremos o esquema a seguir:<br><br>

<center> <img src="fig9.jpg" alt="Logo do IFBA" width="200" height="200"> </center><br>

Dessa forma: $U = \{1,2,3,4,5,6,7,8\}$, $A = \{1,2,4,5\}$, $B = \{2,3,5,6\}$ e $C = \{4,5,6,7\}$.

Note que queremos obter qual dentre os conjuntos presentes nas alternativas corresponde a o conjunto $\{1\}$.

Pois bem,

a) $$
\begin{aligned}
(\overline{A \cup B}) \cup C
&= (\overline{\{1,2,4,5\} \cup \{2,3,5,6\}}) \cup \{4,5,6,7\}\\
&= \overline{\{1,2,3,4,5,6\}} \cup \{4,5,6,7\}\\
&= \{7,8\} \cup \{4,5,6,7\}\\
&= \{4,5,6,7,8\} \neq \{1\}\\
\end{aligned}
$$

b) $$
\begin{aligned}
\overline{B \cup C}
&= \overline{\{2,3,5,6\} \cup \{4,5,6,7\}}\\
&= \overline{\{2,3,4,5,6,7\}}\\
&= \{1,8\} \neq \{1\}\\
\end{aligned}
$$

c) $$
\begin{aligned}
\overline{C} \cap \overline{B} \cap \overline{A}
&= \overline{\{4,5,6,7\}} \cap \overline{\{2,3,5,6\}} \cap \overline{\{1,2,4,5\}}\\
&= \{1,2,3,8\} \cap \{1,4,7,8\} \cap \{3,6,7,8\}\\
&= \{8\} \neq \{1\}\\
\end{aligned}
$$

d) $$
\begin{aligned}
A - B \cup C
&= \{1,2,4,5\} - \{2,3,5,6\} \cup \{4,5,6,7\}\\
&= \{1,2,4,5\} - \{2,3,4,5,6,7\}\\
&= \{1\} \text{ (OK) }\\
\end{aligned}
$$

e) $$
\begin{aligned}
A - A \cap B \cap C
&= \{1,2,4,5\} - \{1,2,4,5\} \cap \{2,3,5,6\} \cap \{4,5,6,7\}\\
&= \{1,2,4,5\} - \{5\}\\
&= \{1,2,4\} \neq \{1\}\\
\end{aligned}
$$

Portanto, a alternativa correta é a "d".


#### Exercício 4 [🎥](https://www.youtube.com/watch?v=W3gP4b7Y7oY)

Em uma prova sobre o corpo humano, havia três questões. A primeira tratava do sistema circulatório, a segunda, do sistema respiratório, e a terceira, do sistema nervoso. Exatamente $29$ alunos fizeram a prova. Sabe-se que:

- $15$ alunos acertaram a primeira questão.
- $7$ alunos acertaram somente a segunda.
- $1$ aluno acertou somente a terceira.
- $11$ alunos acertaram a segunda e a terceira.
- Nenhum aluno errou todas as questões.

Quantos alunos acertaram as três questões?

##### Solução

Sejam $A$, $B$ e $C$ os conjuntos de alunos que acertaram a primeira, a segunda e a terceira questão, respectivamente. Indicamos por $x$ o número de alunos que acertaram as três. Como $11$ acertaram a segunda e a terceira, $11-x$ alunos acertaram essas duas e erraram a primeira.

Para preencher as regiões ligadas à primeira questão, sejam $a=n(A\cap B)$ e $b=n(A\cap C)$. Quem acertou as três questões está contado tanto em $a$ como em $b$. Por isso, nas regiões que correspondem a exatamente duas questões, ficam $a-x$ alunos para a primeira e a segunda e $b-x$ para a primeira e a terceira. Temos o seguinte esquema:<br><br>

<center> <img src="fig10.jpg" alt="Logo do IFBA" width="250" height="200"> </center><br>

Os $15$ alunos que acertaram a primeira estão distribuídos em quatro regiões. Subtraindo as três intersecções já preenchidas, encontramos o número de alunos que acertaram somente a primeira:

$$15-(a-x)-(b-x)-x=15-a-b+x.$$

Agora somamos as quantidades de todas as regiões. A região externa tem $0$ alunos, pois todos acertaram ao menos uma questão.

$$
\begin{aligned}
(15-a-b+x)+(a-x)+(b-x)+x+7+(11-x)+1+0&=29,\\
34-x&=29,\\
x&=5.
\end{aligned}
$$

Portanto, $5$ alunos acertaram as três questões.
