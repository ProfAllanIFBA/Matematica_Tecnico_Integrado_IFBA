## 9.2 Operações com intervalos

Ao representar os intervalos na reta, marque os extremos e destaque o trecho de cada conjunto. Em seguida, selecione os pontos conforme a operação pedida, conferindo também cada extremo.

###
<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;"> <strong>União e intersecção</strong>. Na união $A\cup B$, entram os pontos que pertencem a pelo menos um dos conjuntos. Na intersecção $A\cap B$, entram os pontos que pertencem aos dois conjuntos. </div>

## 9.3 Aplicação a um sistema de inequações

Resolver uma inequação consiste em determinar todos os valores que tornam verdadeira uma desigualdade. Por exemplo, em $3x+10\leq5x+8$, o valor $x=1$ é solução, pois $13\leq13$. Já $x=0$ fica fora do conjunto-solução, pois $10>8$.

Em um sistema de inequações, um valor de $x$ é solução quando satisfaz todas as desigualdades. Podemos resolver cada inequação separadamente e depois determinar a intersecção dos conjuntos-solução.

#### Exemplo 2

Determine, em $\mathbb R$, o conjunto-solução do sistema

$$
\begin{cases}
3x+10\leq5x+8,\\
5x+8\leq4x+20.
\end{cases}
$$

##### Solução

Na primeira inequação, reunimos os termos com $x$ no segundo membro para obter um coeficiente positivo:

$$
\begin{aligned}
3x+10\leq5x+8
&\iff 10-8\leq5x-3x\\
&\iff 2\leq2x\\
&\iff 1\leq x.
\end{aligned}
$$

Assim, $S_1=[1,+\infty[$.

Na segunda inequação, reunimos os termos com $x$ no primeiro membro:

$$
\begin{aligned}
5x+8\leq4x+20
&\iff 5x-4x\leq20-8\\
&\iff x\leq12.
\end{aligned}
$$

Assim, $S_2=]-\infty,12]$. Como as duas condições devem ser satisfeitas ao mesmo tempo, <br><br>

Portanto, $S=S_1\cap S_2=[1,12]$.

$$B=(A\cup B)-(A-B).$$
