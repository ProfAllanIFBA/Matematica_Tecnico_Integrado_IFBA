# Aula 2: Conjunto das partes, união e intersecção [🎥](https://www.youtube.com/watch?v=OQTbAeEDLWU)

## Introdução

Na aula anterior, estudamos as relações de pertinência e de inclusão. Agora vamos reunir todos os subconjuntos de um conjunto e, em seguida, estudar duas operações com conjuntos: a união e a intersecção.

Manteremos a notação $\subseteq$ para a inclusão que admite igualdade. Os símbolos $\in$ e $\notin$ continuam indicando pertinência e não pertinência.

## 2.1 Conjunto das partes de um conjunto

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;"> <strong>Definição</strong>: Dado um conjunto $A$, chamamos de <strong>conjunto das partes de $A$</strong>, indicado por $\mathcal{P}(A)$, o conjunto cujos elementos são todos os subconjuntos de $A$. </div>

Assim,

$$\mathcal{P}(A)=\{X\mid X\subseteq A\}.$$

Isso significa que $X\in\mathcal{P}(A)$ se, e somente se, $X\subseteq A$.

#### Exemplo 1

Determine o conjunto das partes de $A=\{1,2,3\}$.

##### Solução

Vamos organizar os subconjuntos de $A$ pela quantidade de elementos:

i) Nenhum elemento: $\emptyset$;

ii) Um elemento: $\{1\}$, $\{2\}$ e $\{3\}$;

iii) Dois elementos: $\{1,2\}$, $\{1,3\}$ e $\{2,3\}$;

iv) Três elementos: $\{1,2,3\}$.

Portanto,

$$\mathcal{P}(A)=\big\{\emptyset,\{1\},\{2\},\{3\},\{1,2\},\{1,3\},\{2,3\},\{1,2,3\}\big\}.$$

Note que $\mathcal{P}(A)$ tem $8$ elementos, e cada um deles é um conjunto. As chaves externas delimitam $\mathcal{P}(A)$; as chaves internas delimitam seus elementos não vazios.

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;"> <strong>Observação</strong>: Como $\emptyset\subseteq A$ e $A\subseteq A$, temos sempre $\emptyset\in\mathcal{P}(A)$ e $A\in\mathcal{P}(A)$. </div>

No exemplo anterior, $1\in A$ e $\{1\}\subseteq A$. Por isso, $\{1\}\in\mathcal{P}(A)$.

Por outro lado, $1\notin\mathcal{P}(A)$: os elementos de $\mathcal{P}(A)$ são subconjuntos de $A$, e não os números de $A$ isoladamente.

### Quantos elementos tem o conjunto das partes?

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;"> <strong>Propriedade</strong>: Se um conjunto finito $A$ tem $n$ elementos, então $\mathcal{P}(A)$ tem $2^n$ elementos. Indicando por $n(A)$ o número de elementos de $A$, temos $n(\mathcal{P}(A))=2^{n(A)}$. </div>

**Justificativa:** Considere $A=\{a_1,a_2,\ldots,a_n\}$, com $n$ elementos distintos. Para formar um subconjunto $X$ de $A$, temos duas possibilidades para cada elemento $a_{i}$: $a_{i} \in X$ ou $a_{i} \notin X$.

Como as escolhas podem ser feitas independentemente, o número de subconjuntos é

$$\underbrace{2\cdot2\cdot\ldots\cdot2}_{n\text{ fatores}}=2^n.$$

Em particular, para $A=\{1,2,3\}$, obtemos $2^3=8$ subconjuntos, conforme a listagem do Exemplo 1.

**Observação:** O conjunto vazio também possui conjunto das partes:

$$\mathcal{P}(\emptyset)=\{\emptyset\}.$$

De fato, o único subconjunto de $\emptyset$ é o próprio $\emptyset$. Assim, $n(\mathcal{P}(\emptyset))=2^0=1$. Não confunda $\emptyset$, que não possui elementos, com $\{\emptyset\}$, que possui um elemento.

## 2.2 União de conjuntos

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;"> <strong>Definição</strong>: A <strong>união</strong> de dois conjuntos $A$ e $B$, indicada por $A\cup B$, é o conjunto formado pelos elementos que pertencem a $A$ <strong>ou</strong> a $B$. </div>

Escrevemos

$$A\cup B=\{x\mid x\in A\text{ ou }x\in B\}.$$

Lemos $A\cup B$ como “$A$ união $B$”. O “ou” é inclusivo: basta pertencer a pelo menos um dos conjuntos, podendo pertencer aos dois.

Na representação tabular da união, cada elemento é escrito uma única vez.

#### Exemplo 2

Considere os conjuntos

$$A=\{1,2,3\},\quad B=\{1,5\},\quad C=\{1,2,3,4,5\}\quad\text{e}\quad D=\{4,7\}.$$

Determine:

a) $A\cup B$

b) $A\cup C$

c) $A\cup D$

##### Solução

a) Reunimos os elementos de $A$ e de $B$, sem repetir o $1$:

$$A\cup B=\{1,2,3,5\}.$$

b) Todo elemento de $A$ já pertence a $C$. Assim,

$$A\cup C=\{1,2,3,4,5\}=C.$$

c) Os conjuntos $A$ e $D$ não têm elementos em comum. Ao reuni-los, obtemos

$$A\cup D=\{1,2,3,4,7\}.$$

### Como interpretar a união em um diagrama de Venn

Para representar $A\cup B$, destacamos toda a região correspondente a $A$ e toda a região correspondente a $B$, incluindo a parte comum.

Se um conjunto está contido no outro, a união corresponde ao conjunto que contém o outro. Se não há elementos em comum, destacamos as duas regiões separadas. Acompanhe a figura a seguir:<br><br>

A cor cinza foi usada para sinalizar a região correspondente às uniões.

### Propriedades da união

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;"> <strong>Propriedades</strong>: Para quaisquer conjuntos $A$, $B$ e $C$, valem as propriedades a seguir. </div>

i) **Relação com a inclusão:**

$$A\subseteq B\quad\Longleftrightarrow\quad A\cup B=B.$$

Se $A$ está contido em $B$, reunir seus elementos aos de $B$ não acrescenta elementos. Reciprocamente, se $A\cup B=B$, todo elemento de $A$ já pertence a $B$.

Em particular,

$$A\cup A=A\qquad\text{e}\qquad A\cup\emptyset=A.$$

ii) **Comutativa:** A ordem dos conjuntos não altera a união:

$$A\cup B=B\cup A.$$

iii) **Associativa:** Podemos agrupar as uniões de duas maneiras:

$$(A\cup B)\cup C=A\cup(B\cup C).$$

Em ambos os casos, reunimos os elementos que pertencem a pelo menos um dos três conjuntos. Podemos escrever simplesmente $A\cup B\cup C$.

## 2.3 Intersecção de conjuntos

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;"> <strong>Definição</strong>: A <strong>intersecção</strong> de dois conjuntos $A$ e $B$, indicada por $A\cap B$, é o conjunto formado pelos elementos que pertencem a $A$ <strong>e</strong> a $B$, simultaneamente. </div>

Escrevemos

$$A\cap B=\{x\mid x\in A\text{ e }x\in B\}.$$

Lemos $A\cap B$ como “$A$ intersecção $B$”. Para pertencer à intersecção, o elemento precisa pertencer aos dois conjuntos.

#### Exemplo 3

Retome os conjuntos do Exemplo 2:

$$A=\{1,2,3\},\quad B=\{1,5\},\quad C=\{1,2,3,4,5\}\quad\text{e}\quad D=\{4,7\}.$$

Determine:

a) $A\cap B$

b) $A\cap C$

c) $A\cap D$

##### Solução

a) O único elemento que pertence a $A$ e a $B$ é o $1$. Portanto,

$$A\cap B=\{1\}.$$

b) Como $A\subseteq C$, todos os elementos de $A$ pertencem também a $C$. Assim,

$$A\cap C=\{1,2,3\}=A.$$

c) Não há elementos que pertençam a $A$ e a $D$ simultaneamente. Logo,

$$A\cap D=\emptyset.$$

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;"> <strong>Conjuntos disjuntos</strong>: Dois conjuntos são <strong>disjuntos</strong> quando não possuem elementos em comum, isto é, quando sua intersecção é o conjunto vazio. </div>

No exemplo anterior, $A$ e $D$ são disjuntos.

**Observação:** Quando a intersecção é vazia, escrevemos $\emptyset$, e não $\{\emptyset\}$. O conjunto $\{\emptyset\}$ possui um elemento.

### Como interpretar a intersecção em um diagrama de Venn

Para representar $A\cap B$, destacamos somente a região comum aos dois conjuntos.

Se $B\subseteq A$, essa região é todo o conjunto $B$. Se $A$ e $B$ são disjuntos, não existe região comum a destacar. Acompanhe a figura a seguir:<br><br>

### Propriedades da intersecção

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;"> <strong>Propriedades</strong>: Para quaisquer conjuntos $A$, $B$ e $C$, valem as propriedades a seguir. </div>

i) **Relação com a inclusão:**

$$A\subseteq B\quad\Longleftrightarrow\quad A\cap B=A.$$

Se $A$ está contido em $B$, todos os elementos de $A$ são comuns aos dois conjuntos. Reciprocamente, se $A\cap B=A$, todo elemento de $A$ pertence também a $B$.

Em particular,

$$A\cap A=A\qquad\text{e}\qquad A\cap\emptyset=\emptyset.$$

ii) **Comutativa:** A ordem dos conjuntos não altera a intersecção:

$$A\cap B=B\cap A.$$

iii) **Associativa:** Podemos agrupar as intersecções de duas maneiras:

$$(A\cap B)\cap C=A\cap(B\cap C).$$

Em ambos os casos, selecionamos os elementos que pertencem aos três conjuntos simultaneamente. Podemos escrever simplesmente $A\cap B\cap C$.

#### Exercício 1 [🎥](https://youtu.be/z5nJFCbl-8Y?si=m5wjLQnM4oSQBqyV)

Seja $\mathcal{P}(A)$ o conjunto das partes de $A$. Sabendo que $\{a\} \in \mathcal{P}(A)$, $\{x\} \in \mathcal{P}(A)$, $\{m,n\} \in \mathcal{P}(A)$ e que $\mathcal{P}(A)$ possui $16$ elementos, determine o conjunto $A$.

##### Solução

Note que, se $\mathcal{P}(A)$ tem $16$ elementos então, temos que

$$2^{n} = 16$$

em que $n$ é o número de elementos de $A$. Assim,

$$2^{n} = 16 = 2^{4} \Rightarrow n = 4.$$

Dessa forma $A$ possui $4$ elementos.

Por outro lado,

$$
\begin{aligned}
\{a\} \in \mathcal{P}(A)
&\Rightarrow a \in A\\
\{x\} \in \mathcal{P}(A)
&\Rightarrow x \in A\\
\{m,n\} \in \mathcal{P}(A)
&\Rightarrow m,n \in A\\
\end{aligned}
$$

Portanto,

$$A = \{a,x,m,n\}.$$

#### Exercício 6 [🎥](https://youtu.be/pTjMYZwAF-k?si=ZoTCxWLsugnEILlL)

Resolva a equação

$$(3x-1)(2x^{2} - x - 1) = 0$$

no conjunto universo $U = \mathbb{Z}$.

##### Solução

Considere a seguinte propriedade:

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;"> O produto $ab$ de dois inteiros $a$ e $b$ é nulo se, e somente se, $a = 0$ ou $b = 0$. Essa propriedade se estende também aos números reais.</div>

Usando a propriedade acima, temos:

$$(3x-1)(2x^{2} - x - 1) = 0 \Leftrightarrow 3x - 1 = 0 \ \ \text{ou} \ \ 2x^{2} - x - 1 = 0.$$

Assim, basta resolvermos cada uma das equações separadamente:

i) Temos:

$$
\begin{aligned}
3x - 1 = 0
&\Rightarrow 3x = 1\\
&\Rightarrow x = \frac{1}{3}\\
\end{aligned}
$$

ii) Temos:

$$
\begin{aligned}
\Delta = b^{2} - 4ac
&=(-1)^{2} - 4\cdot2\cdot(-1)\\
&=9
\end{aligned}
$$

$$
\begin{aligned}
x = \frac{-b \pm \sqrt{\Delta}}{2a}
&=\frac{-(-1) \pm \sqrt{9}}{2\cdot2}\\
&=\frac{1 \pm 3}{4}\\
&= 1 \text{ ou } -\frac{1}{2}\\
\end{aligned}
$$

Assim, temos que a solução final, considerando $U = \mathbb{Z}$ será dada por:

$$S = \left\{\frac{1}{3},1,-\frac{1}{2}\right\} \cap \mathbb{Z} = \{1\}.$$
