# Aula 7: Números Reais e Números Irracionais [🎥](https://www.youtube.com/watch?v=AUGmIvGzzdE&list=PL8aWdrmfXHiKJeft-RcH9OXXBBgoGdQy7&index=26)

## Introdução

Na Aula 6, vimos que $\sqrt{2}$ é irracional. Em outras palavras, nenhuma fração $\frac{p}{q}$, com $p,q\in\mathbb{Z}$ e $q\neq 0$, é igual a $\sqrt{2}$. Nesta aula, vamos estudar os números que têm essa mesma característica e ver como eles, junto com os racionais, formam o conjunto dos números reais.

## 7.1 O conjunto dos números irracionais

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Definição</strong>. Um número real é irracional quando não pode ser escrito na forma $\frac{p}{q}$, com $p,q\in\mathbb{Z}$ e $q\neq 0$. Denotaremos o conjunto dos irracionais por $\overline{\mathbb{Q}}$ ou $\mathbb{Q}'$. Esse conjunto é o complementar de $\mathbb{Q}$ em $\mathbb{R}$.
</div>

A representação decimal de um número irracional é infinita e não periódica. Dois exemplos são

$$
\sqrt{2}=1{,}41421356\ldots \qquad \text{e} \qquad \pi=3{,}14159265\ldots
$$

As casas decimais exibidas dão uma ideia do valor de cada número. Para decidir se uma representação infinita é periódica, é preciso considerar a expansão inteira. Nos números racionais, a representação decimal usual é finita ou é uma dízima periódica.

### Raízes de números naturais

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Propriedade</strong>. Se $a\in\mathbb{N}$ e $n$ é um inteiro maior ou igual a $2$, então $\sqrt[n]{a}$ é um número inteiro ou irracional. Em particular, ela é inteira quando $a$ é a $n$-ésima potência de um número natural.
</div>

**Justificativa**. Suponha que $\sqrt[n]{a}=\frac{p}{q}$ seja uma fração irredutível, com $p\geq 0$ e $q>0$. Elevando os dois membros à potência $n$, obtemos $p^n=aq^n$. Se $q>1$, algum primo que divide $q$ também divide $p^n$ e, portanto, divide $p$. Isso contradiz o fato de a fração ser irredutível. Logo, $q=1$ e $\sqrt[n]{a}=p$ é inteiro.

#### Exemplo 1

Classifique $\sqrt{49}$, $\sqrt{8}$ e$\sqrt[3]{27}$ como inteiros ou irracionais.

##### Solução

a) Como $7^2=49$, temos $\sqrt{49}=7\in\mathbb{Z}$.

b) Como $2^2<8<3^2$, obtemos $2<\sqrt{8}<3$. Entre esses dois inteiros consecutivos não há outro inteiro. Pela propriedade acima, $\sqrt{8}$ é irracional.

c) Como $3^3=27$, temos $\sqrt[3]{27}=3\in\mathbb{Z}$.

## 7.2 Operações com números racionais e irracionais

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Soma e diferença</strong>. Se $r\in\mathbb{Q}$ e $i\in\overline{\mathbb{Q}}$, então $r+i$, $r-i$ e $i-r$ são irracionais.
</div>

##### Demonstração

Suponha, por absurdo, que $r+i=s$, com $s\in\mathbb{Q}$. Então

$$i=s-r.$$

Como a diferença de dois racionais é racional, teríamos $i\in\mathbb{Q}$, uma contradição. Portanto, $r+i\in\overline{\mathbb{Q}}$.

O mesmo raciocínio vale para as diferenças. Se $r-i=s$ fosse racional, então $i=r-s$ seria racional. Se $i-r=s$ fosse racional, então $i=s+r$ também seria racional.

#### Exemplo 2

Classifique $2+\sqrt{2}$ e $\sqrt{2}-1$ como racionais ou irracionais.

##### Solução

O número $2$ é inteiro e, portanto, racional. Como $\sqrt{2}$ é irracional, a soma $2+\sqrt{2}$ é irracional. Do mesmo modo, $\sqrt{2}-1$ é irracional, pois subtraímos um racional de um irracional. Nesse caso, a representação decimal é $0{,}41421356\ldots$.

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Produto e quociente</strong>. Se $r\in\mathbb{Q}$, $r\neq 0$ e $i\in\overline{\mathbb{Q}}$, então $r\cdot i$, $\frac{i}{r}$ e $\frac{r}{i}$ são irracionais.
</div>

Se o produto $r\cdot i=s$ fosse racional, então $i=\frac{s}{r}$ seria racional, pois $r\neq 0$. Isso contradiz a hipótese.

Para o quociente $\frac{i}{r}$, basta observar que $\frac{i}{r}=i\cdot\frac{1}{r}$, um produto de um irracional por um racional diferente de zero. Já se $\frac{r}{i}=s$ fosse racional, teríamos $s\neq 0$, pois $r\neq 0$, e $i=\frac{r}{s}$ seria racional.

**Observação**. A condição $r\neq 0$ é importante. O produto $0\cdot i$ e o quociente $\frac{0}{i}$ são iguais a $0$, que é racional. O quociente $\frac{i}{0}$ é indefinido. Todo irracional é diferente de zero.

#### Exemplo 3

Classifique $2\sqrt{2}$ e $\frac{6}{\sqrt{12}}$ como racionais ou irracionais.

##### Solução

a) Como $2=\sqrt{4}$, podemos escrever

$$2\sqrt{2}=\sqrt{4}\cdot\sqrt{2}=\sqrt{8}.$$

Como $2^2<8<3^2$, a raiz $\sqrt{8}$ não é inteira e, portanto, é irracional.

b) Racionalizando o denominador, temos

$$
\frac{6}{\sqrt{12}}
=\frac{6\sqrt{12}}{12}
=\frac{\sqrt{12}}{2}
=\frac{\sqrt{12}}{\sqrt{4}}
=\sqrt{3}.
$$

Como $1^2<3<2^2$, concluímos que $\sqrt{3}$ é irracional.

### Operações entre dois irracionais

Uma operação entre dois irracionais pode dar um resultado racional. Veja alguns exemplos.

i) **Soma**. A soma de um irracional com seu oposto é racional.

$$\sqrt{2}+(-\sqrt{2})=0\in\mathbb{Q}.$$

ii) **Produto**. O produto de $\sqrt{2}$ por ele mesmo é racional.

$$\sqrt{2}\cdot\sqrt{2}=\sqrt{4}=2\in\mathbb{Q}.$$

Também há casos em que o resultado é irracional. Por exemplo, $\sqrt{2}+\sqrt{2}=2\sqrt{2}$ e $\sqrt{2}\cdot\sqrt{3}=\sqrt{6}$ são irracionais. Assim, o conjunto dos irracionais não é fechado para a soma nem para o produto.

## 7.3 O conjunto dos números reais

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Definição</strong>. O conjunto dos números reais, representado por $\mathbb{R}$, é a união do conjunto dos números racionais com o conjunto dos números irracionais.
</div>

$$\mathbb{R}=\mathbb{Q}\cup\overline{\mathbb{Q}}.$$

Os naturais e os inteiros já estão contidos nos racionais. Podemos escrever

$$\mathbb{N}\subseteq\mathbb{Z}\subseteq\mathbb{Q}\subseteq\mathbb{R}
   \qquad\text{e}\qquad \overline{\mathbb{Q}}\subseteq\mathbb{R}.$$
  
Temos, portanto, o esquema a seguir:<br><br>
   
   <center> <img src="fig1.jpg" width="350"> </center>

Cada ponto da reta numérica corresponde a um número real. Os reais permitem representar, por exemplo, contagens, comprimentos e outras medidas.

### Fechamento das operações

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Propriedade</strong>. Se $a,b\in\mathbb{R}$, então $a+b$, $a-b$ e $a\cdot b$ pertencem a $\mathbb{R}$. Se, além disso, $b\neq 0$, então $\frac{a}{b}\in\mathbb{R}$.
</div>

Por exemplo,

$$
\begin{aligned}
  3+\sqrt{3} &\in \mathbb{R},\\
  2 + 5 & \in \mathbb{R},\\
  \sqrt{2}+\sqrt{7} &\in \mathbb{R},\\
  3-\sqrt{3} &\in \mathbb{R},\\
  -\sqrt{2} - \sqrt{7} &\in \mathbb{R},\\
  2\sqrt{5} & \in \mathbb{R},\\
  \sqrt{3}\cdot\sqrt{7} & \in \mathbb{R},\\
  -4\cdot\frac{1}{\sqrt{3}} & \in \mathbb{R},\\
  \frac{\sqrt{5}}{\sqrt{2}} & \in \mathbb{R}.
\end{aligned}
$$

### Oposto e inverso

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Oposto</strong>. Para cada $x\in\mathbb{R}$, existe um único $y\in\mathbb{R}$ tal que $x+y=0$. Esse número é o oposto de $x$ e é indicado por $-x$.
</div>

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Inverso</strong>. Para cada $x\in\mathbb{R}$ com $x\neq 0$, existe um único $y\in\mathbb{R}$ tal que $x\cdot y=1$. Esse número é o inverso de $x$ e é indicado por $\frac{1}{x}$ ou $x^{-1}$.
</div>

Por exemplo, o oposto de $2$ é $-2$, pois $2+(-2)=0$. O inverso de $2$ é $\frac{1}{2}$, pois $2\cdot\frac{1}{2}=1$. O oposto de $0$ é o próprio $0$. Como $0\cdot y=0$ para todo $y$ real, o zero não possui inverso multiplicativo.

## Exercícios Resolvidos

#### Exercício 1 [🎥](https://www.youtube.com/watch?v=HAW4PVBQWoc&list=PL8aWdrmfXHiKJeft-RcH9OXXBBgoGdQy7&index=27)

(FATEC) Sejam $a$ e $b$ dois números irracionais. Analise as afirmações a seguir.

I. $a\cdot b$ é irracional.

II. $a+b$ é irracional.

III. $a-b$ pode ser racional.

Assinale a alternativa correta.

a) As três afirmações são falsas.

b) As três afirmações são verdadeiras.

c) Somente I e III são verdadeiras.

d) Somente I é verdadeira.

e) Somente I e II são falsas.

##### Solução

**I. Falsa.** Tome $a=b=\sqrt{2}$. Apesar de $a$ e $b$ serem irracionais, o produto é

$$a\cdot b=\sqrt{2}\cdot\sqrt{2}=2\in\mathbb{Q}.$$

**II. Falsa.** Tome $a=\sqrt{2}$ e $b=1-\sqrt{2}$. Ambos são irracionais, mas

$$a+b=\sqrt{2}+(1-\sqrt{2})=1\in\mathbb{Q}.$$

**III. Verdadeira.** Basta encontrar um caso em que a diferença seja racional. Tome $a=\sqrt{2}$ e $b=1+\sqrt{2}$. A soma $1+\sqrt{2}$ é irracional e

$$a-b=\sqrt{2}-(1+\sqrt{2})=-1\in\mathbb{Q}.$$

Portanto, a resposta é a **alternativa e)**.

#### Exercício 2 [🎥](https://www.youtube.com/watch?v=4tCLMSObH4E&list=PL8aWdrmfXHiKJeft-RcH9OXXBBgoGdQy7&index=28)

(Fuvest 1991) Na figura a seguir estão representados os números reais $0$, $x$, $y$ e $1$:<br><br>
   
   <center> <img src="fig2.jpg" width="180"> </center>

Qual a posição do número $xy$?

a) À esquerda de $0$.

b) Entre $0$ e $x$.

c) Entre $x$ e $y$.

d) Entre $y$ e $1$.

e) À direita de $1$.

##### Solução

A ordem dos pontos na reta dá $0<x<y<1$. Em particular, $0<y<1$. Como $x>0$, podemos multiplicar os três membros dessa desigualdade por $x$ sem alterar o sentido das desigualdades. Assim,

$$0\cdot x < y\cdot x < 1\cdot x,$$

ou seja,

$$0<xy<x.$$

Portanto, $xy$ está entre $0$ e $x$. A resposta é a **alternativa b)**.

#### Exercício 3 [🎥](https://www.youtube.com/watch?v=m2JSxDb4ylg&list=PL8aWdrmfXHiKJeft-RcH9OXXBBgoGdQy7&index=29)

Seja $a=\sqrt{2}$. Mostre que, para $n\in\mathbb{N}$,

a) se $n$ for par, então $a^n$ é racional;

b) se $n$ for ímpar, então $a^n$ é irracional.

##### Solução

a) Se $n$ é par, podemos escrever $n=2k$, com $k\in\mathbb{N}$. Portanto,

$$
a^n=(\sqrt{2})^{2k}
   =\bigl((\sqrt{2})^2\bigr)^k
   =2^k.
$$

Como $2^k$ é um número natural, $a^n$ é racional. Isso inclui o caso $n=0$, em que $a^0=1$.

b) Se $n$ é ímpar, escrevemos $n=2k+1$, com $k\in\mathbb{N}$. Assim,

$$
a^n=(\sqrt{2})^{2k+1}
   =(\sqrt{2})^{2k}\cdot\sqrt{2}
   =2^k\sqrt{2}.
$$

Como $2^k$ é racional e diferente de zero, e $\sqrt{2}$ é irracional, o produto $2^k\sqrt{2}$ é irracional.

#### Exercício 4 [🎥](https://www.youtube.com/watch?v=SSmiiflvgDg&list=PL8aWdrmfXHiKJeft-RcH9OXXBBgoGdQy7&index=30)

Resolva, no conjunto dos irracionais, a equação

$$x^2-(1+\sqrt{2})x+\sqrt{2}=0.$$

##### Solução

Vamos resolver a equação em $\mathbb{R}$ e verificar quais raízes pertencem a $\overline{\mathbb{Q}}$. Pela fórmula do segundo grau, teríamos de calcular a raiz quadrada do discriminante

$$\Delta=(1+\sqrt{2})^2-4\sqrt{2}=3-2\sqrt{2}.$$

Podemos reconhecer as raízes pela soma e pelo produto.

Em uma equação quadrática $ax^2+bx+c=0$ com duas raízes reais $x_1$ e $x_2$, a forma fatorada é

$$
a(x-x_1)(x-x_2)
   = a\bigl[x^2-(x_1+x_2)x+x_1x_2\bigr].
$$

Comparando os coeficientes, obtemos $x_1+x_2=-\frac{b}{a}$ e $x_1x_2=\frac{c}{a}$. Para a equação do enunciado, procuramos dois números cuja soma e cujo produto sejam, respectivamente,

$$x_1+x_2=1+\sqrt{2}\qquad\text{e}\qquad x_1x_2=\sqrt{2}.$$

Observe que $1$ e $\sqrt{2}$ satisfazem essas duas condições. De fato,

$$
\begin{aligned}
x^2-(1+\sqrt{2})x+\sqrt{2}
  &=x^2-x-\sqrt{2}x+\sqrt{2}\\
  &=(x-1)(x-\sqrt{2}).
\end{aligned}
$$

Como o produto deve ser zero, temos $x=1$ ou $x=\sqrt{2}$. Como $1$ é racional e $\sqrt{2}$ é irracional, o conjunto solução no universo pedido é

$$S=\left\{\sqrt{2}\right\}.$$
