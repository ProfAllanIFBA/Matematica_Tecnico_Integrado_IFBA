## Introdução

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
Um conjunto expressa a ideia de uma coleção de objetos. Cada objeto da coleção é um elemento. Se um elemento $a$ pertence a um conjunto $X$ escrevemos $a \in X$. Do contrário, escrevemos $a \notin X$.
</div>

Se $B$ é o conjunto dos planetas do Sistema Solar, a Terra é um de seus elementos. Escrevemos $\text{Terra}\in B$ e lemos “a Terra pertence a $B$”.

## 1.1 Como representar um conjunto

Adotaremos três formas básicas de representação de um conjunto. Tomemos como exemplo o conjunto formado pelas vogais da língua portuguesa.

###
<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;"> <strong>Representação Tabular</strong>: Usamos chaves para separar os elementos. </div>


$$A = \{a,e,i,o,u\}.$$

Em particular, $a \in A$ e $b \notin A$.

###
<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;"> <strong>Representação por Diagrama de Venn</strong>: Os elementos do conjunto ficam internos a uma curva fechada. </div>

###
<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;"> <strong>Representação por Propriedade</strong>: Uma propriedade é usada no lugar de explicitar os elementos. </div>

$$D = \{1,2,3,6\}.$$

## 1.2 Conjuntos particulares e universo

###
<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;"> <strong>Unitário</strong>: Conjunto que possui exatamente um elemento. </div>

$$E = \{x | x \ \text{é estrela do Sistema Solar} \}.$$

###
<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;"> <strong>Vazio ($\emptyset$)</strong>: Conjunto que não possui elemento algum. </div>

$$\emptyset = \{x \in \mathbb{N} | 0 \cdot x = 5\}.$$

O conjunto vazio $\emptyset$ também pode ser representado por $\{ \}$.

###
<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;"> <strong>Finito</strong>: É possível terminar a contagem de seus elementos.</div>

$$A = \{x| x \ \text{é vogal da língua portuguesa}\}.$$

###
<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;"> <strong>Infinito</strong>: Sempre há outros elementos a considerar.</div>

$$\mathbb{N} = \{0,1,2,3,...\} \ \ \text{e} \ \ \mathbb{Z} = \{...,-2,-1,0,1,2,...\}.$$

**Observação**: Alguns autores não consideram o $0$ natural. Neste material estamos considerando $0$ natural.

###
<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;"> <strong>Universo (U)</strong>: Conjunto que reúne todos os elementos admitidos em um estudo.</div>

Considere as soluções da equação $x^2=4$:

i) Em $U=\mathbb{Z}$, as soluções são $-2$ e $2$;

ii) Em $U=\mathbb{N}$, apenas $2$ satisfaz as duas condições: ser natural e ter quadrado igual a $4$.

###
<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;"> <strong>Propriedades</strong>: Para todo conjunto $A$, os conjunto vazio $\emptyset$ e o universo $U$ são tais que:

i) $\emptyset \subset A$
    
ii) $A \subset U$
</div>

## 1.3 Quando um conjunto está contido em outro?

###
<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;"> <strong>Definição</strong>: Dizemos que $C$ é um subconjunto de $A$ quando todo elemento de $C$ pertence a $A$. Escrevemos $C\subseteq A$. Para mostrar que a inclusão falha, basta encontrar um elemento de $C$ que esteja fora de $A$. Neste caso, $C \not\subset A$.</div>

Adotaremos $\subseteq$ para a inclusão que admite igualdade. A escrita $A\supseteq C$ expressa a mesma relação, lida na direção oposta: “$A$ contém $C$”.

#### Exemplo 1

Considere os conjuntos $A = \{a,e\}$, $B = \{a,e,i,o,u\}$ e $D = \{a,b\}$. Analise cada uma das inclusões a seguir as julgando como verdadeira ou falsa:

a) $A \subset B$

b) $D \subset B$

##### Solução:

a) Note que todo elemento de $A$ é também um elemento de $B$. Portanto, a inclusão em questão é verdadeira.

b) Note que nem todo elemento de $D$ está contido em $B$ que é o caso de $b$. Dessa forma, a inclusão em questão é falsa. O correto seria $D \not\subset B$.

#### Exemplo 2

Considere o conjunto $A = \{1,2,3,6\}$. Analise cada uma das afirmações a seguir as julgando como verdadeira ou falsa:

a) $1 \in A$

b) $1,4 \in A$

c) $\{1\} \notin A$

d) $\{1\} \in \{\{1\},\{2\},\{3\},\{6\}\}$

e) $1 \subset A$

f) $\{1\} \subset A$

g) $\{1\} \not\subset \{\{1\},\{2\},\{3\},\{6\}\}$

h) $\{\emptyset\} \in A$

i) $\{\emptyset\} \subset A$

##### Solução

a) $V$, b) $F$, c) $V$, d) $V$, e) $F$, f) $V$, g) $V$, h) $F$, i) $F$

## Exercícios Resolvidos

#### Exercício 1 [🎥](https://youtu.be/qfECB4l_XUQ?si=rYuXgyCg86MtXiFL)

Represente os conjuntos na forma tabular:

a) $A = \{x \in \mathbb{N} | 3 \leq x < 10 \}$

b) $B = \{x \in \mathbb{Z} | x^{2} + x - 6 = 0\}$

c) $C = \{x \in \mathbb{N} | x^{2} - 4 = 0\}$

##### Solução

a) É facil ver que:

$$A = \{x \in \mathbb{N} | 3 \leq x < 10 \} = \{3,4,5,6,7,8,9\}.$$

b) Primeiramente devemos resolver a equação polinomial do segundo grau $x^{2} + x - 6 = 0$. Pois bem,

$$
\begin{aligned}
\Delta = b^{2} - 4ac
&=1^{2} - 4\cdot1\cdot(-6)\\
&=25
\end{aligned}
$$

$$
\begin{aligned}
x = \frac{-b \pm \sqrt{\Delta}}{2a}
&=\frac{-1 \pm \sqrt{25}}{2\cdot1}\\
&=\frac{-1 \pm 5}{2}\\
&= -3 \text{ ou } 2\\
\end{aligned}
$$

Assim,

$$B = \{x \in \mathbb{Z} | x^{2} + x - 6 = 0\} = \{-3,2\}.$$

c) Analogamente ao item anterior, devemos resolver a equação polinomial do segundo grau $x^{2} - 4 = 0$. Sendo esta incompleta (termo $b = 0$) fica mais fácil resolver. Pois bem,

$$
\begin{aligned}
x^{2} - 4 = 0
&\Rightarrow x^{2} = 4\\
&\Rightarrow x = \pm \sqrt{2}\\
&\Rightarrow x = -2 \text{ ou } x = 2
\end{aligned}
$$

Contudo, $-2 \notin \mathbb{N}$. Portanto,

$$C = \{x \in \mathbb{N} | x^{2} - 4 = 0\} = \{2\}.$$



#### Exercício 2 [🎥](https://www.youtube.com/watch?v=LRbRST74i54&list=PL8aWdrmfXHiKJeft-RcH9OXXBBgoGdQy7&index=3&pp=iAQBsAgC)

Classifique em $V$ ou $F$ cada uma das afirmações a seguir:

a) $\{x \in \mathbb{N} | 6 \text{ divide } x\} \subset \{x \in \mathbb{N} | 3 \text{ divide } x\}$

b) $\{x \in \mathbb{Z} | x < 10\} \subset \{x \in \mathbb{N} | x > 0\}$

c) $\{x \in \mathbb{Z} | -5 \leq x \leq 8\} \supset \{x \in \mathbb{N} | x < 0\}$

##### Solução

a) Temos que

$$
\begin{aligned}
\{x \in \mathbb{N} | 6 \text{ divide } x\} = \{0,6,12,18,...\}\\
\end{aligned}
$$

e

$$
\begin{aligned}
\{x \in \mathbb{N} | 3 \text{ divide } x\} = \{0,3,6,9,12,...\}\\
\end{aligned}
$$

Note que

$$\{0,6,12,18,...\} \subset \{0,3,6,9,12,...\}$$

é uma inclusão verdadeira.

b) Temos que

$$
\begin{aligned}
\{x \in \mathbb{Z} | x < 10\} = \{...,-2,-1,0,1,...,9\}\\
\end{aligned}
$$

e

$$
\begin{aligned}
\{x \in \mathbb{N} | x > 0\} = \{1,2,3,4,....\}\\
\end{aligned}
$$

Note que

$$\{\boxed{...,-2,-1,0},1,...,9\} \subset \{1,2,3,4,....\}$$

é uma inclusão falsa.

c) Temos que $\{x \in \mathbb{Z} | -5 \leq x \leq 8\} \supset \{x \in \mathbb{N} | x < 0\}$

$$
\begin{aligned}
\{x \in \mathbb{Z} | -5 \leq x \leq 8\} = \{-5,-4,-3,-2,-1,0,1,2,...,8\}\\
\end{aligned}
$$

e

$$
\begin{aligned}
\{x \in \mathbb{N} | x < 0\} = \emptyset\\
\end{aligned}
$$

Note que

$$\{-5,-4,-3,-2,-1,0,1,2,...,8\} \supset \emptyset$$

é uma inclusão verdadeira.

#### Exercício 3 [🎥](https://youtu.be/B6xg3E34c28?si=YvhEHDKN_Kb2asPQ)

Considere o conjunto

$$A = \{1,2,3,4,5,6,7\}.$$

a) Quantos são os subconjuntos de $A$ que não figuram o número $5$?

b) Quantos são os subconjuntos de $A$ que figuram o número $2$ e não figuram o número $5$?

##### Solução

a) Dado um elemento qualquer $a \in A$ e um subconjunto $B$ de $A$ temos duas hipóteses: ou $a \in B$ ou $a \notin B$. Dessa forma, para todo valor $1,2,3,4,6,7$ (todos exceto o $5$), temos dois casos. Por outro lado, para o elemento $5$ temos só um caso pois ele não deve pertencer a subconjunto algum. Assim, o número total de casos é:

$$2 \cdot 2 \cdot 2 \cdot 2 \underset{5\notin A}{\underbrace{\cdot 1}} \cdot 2 \cdot 2 = 2^{6} = 64.$$

Temos, portanto $64$ subconjuntos de $A$ nos quais $5$ não figura.

b) Analogamente ao item anterior, temos que os elementos $1,3,4,6,7$ podem se apresentar de duas formas, estando ou não em um dado subconjunto. Por outro lado, $2$ e $5$ só podem se apresentar de uma forma, $2$ figurando e $5$ não figurando. Assim, o número total de casos é:

$$2 \cdot \underset{2\in A}{\underbrace{\cdot 1}} \cdot 2 \cdot 2 \underset{5\notin A}{\underbrace{\cdot 1}} \cdot 2 \cdot 2 = 2^{5} = 32.$$

Temos, portanto $32$ subconjuntos de $A$ no quais o $2$ figura e o $5$ não figura.
