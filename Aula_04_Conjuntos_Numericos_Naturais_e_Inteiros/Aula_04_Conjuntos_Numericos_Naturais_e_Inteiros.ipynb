# Aula 4: Conjuntos numéricos — números naturais e números inteiros [🎥](https://www.youtube.com/watch?v=88iktMf7gGs&list=PL8aWdrmfXHiKJeft-RcH9OXXBBgoGdQy7&index=17)

## Introdução

Nas aulas anteriores, estudamos relações e operações entre conjuntos. Agora vamos aplicar essas ideias a dois conjuntos numéricos: o dos números naturais e o dos números inteiros. Vamos observar quais operações produzem resultados que permanecem em cada conjunto e como o universo escolhido interfere na solução de uma equação.

## 4.1 Conjunto dos números naturais

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Conjunto dos números naturais.</strong> Os números naturais são usados para contar. Neste material, como na Aula 1, consideramos o zero um número natural.
</div>

Indicamos o conjunto dos números naturais por $\mathbb{N}$. O asterisco em $\mathbb{N}^{*}$ indica que retiramos o zero.

$$
\begin{aligned}
\mathbb{N} &= \{0,1,2,3,\ldots\},\\
\mathbb{N}^{*} &= \mathbb{N}-\{0\}=\{1,2,3,\ldots\}.
\end{aligned}
$$

Alguns autores adotam a convenção de começar a lista dos naturais em $1$. Aqui, o primeiro natural é $0$.

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Sucessor e antecessor.</strong> O sucessor de um natural é obtido acrescentando uma unidade. Para obter o antecessor, retiramos uma unidade, desde que o resultado ainda seja natural.
</div>

Se $n\in\mathbb{N}$, então $n+1$ é seu sucessor. Se $n>0$, então $n-1$ é seu antecessor. Assim, $3$ é o sucessor de $2$ e $2$ é o antecessor de $3$. Esses dois números são **consecutivos**. O zero não tem antecessor em $\mathbb{N}$, pois $0-1=-1\notin\mathbb{N}$.

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Fechamento.</strong> Dizemos que um conjunto é fechado para uma operação quando, ao realizá-la com elementos desse conjunto, o resultado sempre pertence ao próprio conjunto.
</div>

A **soma** e o **produto** de dois naturais são sempre naturais. Por exemplo,

$$
2+3=5\in\mathbb{N}
\qquad\text{e}\qquad
2\cdot 4=8\in\mathbb{N}.
$$

A subtração e a divisão por um natural diferente de zero podem produzir resultados fora de $\mathbb{N}$. Basta um contraexemplo para mostrar que o conjunto não é fechado para uma dessas operações.

$$
2-3=-1\notin\mathbb{N}
\qquad\text{e}\qquad
2\div 4=0{,}5\notin\mathbb{N}.
$$

## 4.2 Conjunto dos números inteiros

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Conjunto dos números inteiros.</strong> Além dos números naturais, os inteiros incluem os números negativos −1, −2, −3 e assim por diante.
</div>

O conjunto dos números inteiros é indicado por $\mathbb{Z}$.

$$\mathbb{Z}=\{\ldots,-3,-2,-1,0,1,2,3,\ldots\}.$$

Os números negativos permitem representar, por exemplo, um saldo devedor. Como todo natural é inteiro, $\mathbb{N}\subseteq\mathbb{Z}$. A inclusão é própria, pois $-1\in\mathbb{Z}$ e $-1\notin\mathbb{N}$.

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Subconjuntos de inteiros.</strong> Usamos os sinais de mais e de menos para indicar o lado do zero considerado e o asterisco para excluir o zero.
</div>

- **Inteiros não nulos** — $\mathbb{Z}^{*}=\mathbb{Z}-\{0\}=\{\ldots,-3,-2,-1,1,2,3,\ldots\}$.
- **Inteiros não negativos** — $\mathbb{Z}_{+}=\{x\in\mathbb{Z}\mid x\geq 0\}=\{0,1,2,3,\ldots\}$.
- **Inteiros não positivos** — $\mathbb{Z}_{-}=\{x\in\mathbb{Z}\mid x\leq 0\}=\{\ldots,-3,-2,-1,0\}$.
- **Inteiros positivos** — $\mathbb{Z}_{+}^{*}=\{x\in\mathbb{Z}\mid x>0\}=\{1,2,3,\ldots\}$.
- **Inteiros negativos** — $\mathbb{Z}_{-}^{*}=\{x\in\mathbb{Z}\mid x<0\}=\{\ldots,-3,-2,-1\}$.

Nesta convenção, $\mathbb{Z}_{+}=\mathbb{N}$ e $\mathbb{Z}_{+}^{*}=\mathbb{N}^{*}$. O zero pertence a $\mathbb{Z}_{+}$ e a $\mathbb{Z}_{-}$.

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Propriedades.</strong> O conjunto dos números inteiros é fechado para a adição, a subtração e a multiplicação.
</div>

Se $a,b\in\mathbb{Z}$, então $a+b$, $a-b$ e $a\cdot b$ também pertencem a $\mathbb{Z}$. Por exemplo,

$$
3+(-2)=1,\qquad 3-7=-4,\qquad (-3)\cdot(-2)=6.
$$

Uma diferença como $3-7$ pode ser calculada em $\mathbb{Z}$ porque $-4$ é inteiro. Para a divisão, o contraexemplo anterior também vale em $\mathbb{Z}$, pois $2\div4=0{,}5\notin\mathbb{Z}$.

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Sucessor e antecessor em ℤ.</strong> Todo inteiro tem sucessor e antecessor no próprio conjunto.
</div>

Para todo $n\in\mathbb{Z}$, seu sucessor é $n+1$ e seu antecessor é $n-1$. Em particular, $-1$ é o antecessor de $0$ em $\mathbb{Z}$ e $0$ é o sucessor de $-1$.

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Oposto de um inteiro.</strong> O oposto de um inteiro é o número que, somado a ele, resulta em zero.
</div>

O oposto de $x$ é $-x$, pois $x+(-x)=0$. Por exemplo, $3$ e $-3$ são opostos um do outro, já que $3+(-3)=0$. O zero é o próprio oposto.

## 4.3 Conjunto universo e conjunto solução

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Conjunto solução.</strong> Ao resolver uma equação em um conjunto universo, só incluímos na solução os elementos desse universo que satisfazem a igualdade.
</div>

#### Exemplo 1

Resolva a equação $x+3=2$ em $\mathbb{N}$. O conjunto solução muda se o universo considerado for $\mathbb{Z}$?

##### Solução

Subtraindo $3$ dos dois membros e, em seguida, dividindo ambos por $3$, obtemos

$$
\begin{aligned}
x + 3 = 2
&\Rightarrow x = 3-2 \\
&\Rightarrow x = -1
\end{aligned}
$$

O único valor que pode satisfazer a igualdade é $-1$. Como ele não é natural nem inteiro, nos dois universos o conjunto solução é $S=\emptyset$.

#### Exemplo 2

Resolva em $\mathbb{Z}$ a equação $2x+4=-2$.

##### Solução

Subtraindo $4$ dos dois membros e dividindo ambos por $2$, obtemos


$$
\begin{aligned}
2x + 4 = -2
&\Rightarrow 2x = -2-4 \\
&\Rightarrow 2x = -6 \\
&\Rightarrow x = - \frac{6}{2}\\
&\Rightarrow x = -3
\end{aligned}
$$

Como $-3\in\mathbb{Z}$ e $2\cdot(-3)+4=-2$, temos $S=\{-3\}$ em $\mathbb{Z}$. Se o universo fosse $\mathbb{N}$, teríamos $S=\emptyset$, pois $-3\notin\mathbb{N}$.

## Exercícios Resolvidos

#### Exercício 1 [🎥](https://www.youtube.com/watch?v=u-BBjVEvUOM&list=PL8aWdrmfXHiKJeft-RcH9OXXBBgoGdQy7&index=19)

Resolva em $\mathbb{N}$ o seguinte sistema de inequações.

$$
\begin{cases}
2x+1>3 & \text{(I)}\\
x+3<10 & \text{(II)}
\end{cases}
$$

##### Solução

**(I)** Subtraindo $1$ de ambos os membros e dividindo por $2$, obtemos

$$
\begin{aligned}
2x+1 > 3
&\Rightarrow 2x > 2 \\
&\Rightarrow x > \frac{2}{2}\\
&\Rightarrow x > 1\\
\end{aligned}
$$

Assim, o conjunto solução da primeira inequação em $\mathbb{N}$ é $A=\{2,3,4,5,\ldots\}$.

**(II)** Subtraindo $3$ dos dois membros, obtemos

$$
\begin{aligned}
x + 3 < 10
&\Rightarrow x < 10-3 \\
&\Rightarrow x < 7
\end{aligned}
$$

Logo, o conjunto solução da segunda inequação em $\mathbb{N}$ é $B=\{0,1,2,3,4,5,6\}$.

Como as duas condições devem ser satisfeitas ao mesmo tempo, a solução do sistema é

$$
\begin{aligned}
S = A \cap B
&= \{2,3,4,5,\ldots\} \cap \{0,1,2,3,4,5,6\}\\
&= \{2,3,4,5,6\}
\end{aligned}
$$

Portanto, a solução do sistema de inequações é dada por:

$$S = \{2,3,4,5,6\}.$$


#### Exercício 2 [🎥](https://www.youtube.com/watch?v=3lXLYpUYCc0&list=PL8aWdrmfXHiKJeft-RcH9OXXBBgoGdQy7&index=20)

Resolva em $\mathbb{Z}$ a seguinte equação.

$$2x^{2}=3-5x.$$

##### Solução

Escrevemos a equação na forma $ax^2+bx+c=0$.

$$2x^2+5x-3=0.$$

Com $a=2$, $b=5$ e $c=-3$, temos

$$
\begin{aligned}
\Delta = b^{2} - 4ac
&=5^{2}-4\cdot2\cdot(-3)\\
&=49
\end{aligned}
$$

$$
\begin{aligned}
x = \frac{-b \pm \sqrt{\Delta}}{2a}
&=\frac{-5 \pm \sqrt{49}}{2\cdot2}\\
&=\frac{-5 \pm 7}{4}\\
&= \frac{1}{2} \text{ ou } -3\\
\end{aligned}
$$

Como $\frac{1}{2}\notin\mathbb{Z}$ e $-3\in\mathbb{Z}$, o conjunto solução em $\mathbb{Z}$ é $S=\{-3\}$.

#### Exercício 3 [🎥](https://www.youtube.com/watch?v=8ba88RSmk_U&list=PL8aWdrmfXHiKJeft-RcH9OXXBBgoGdQy7&index=21)

Um inteiro $n$ é **par** quando existe $k\in\mathbb{Z}$ tal que $n=2k$. Ele é **ímpar** quando existe $k\in\mathbb{Z}$ tal que $n=2k+1$.

Com base nessas definições, demonstre as duas afirmações.

**(I)** A soma de dois números pares é um número par.

**(II)** O produto de dois números ímpares é um número ímpar.

##### Solução

Por exemplo, $8=2\cdot4$ é par, e $7=2\cdot3+1$ e $9=2\cdot4+1$ são ímpares. Nas demonstrações a seguir, usamos as mesmas formas para inteiros quaisquer.

**(I)** Sejam $m$ e $n$ dois inteiros pares. Pela definição, existem $k_{1},k_{2}\in\mathbb{Z}$ tais que $m=2k_{1}$ e $n=2k_{2}$. Então,

$$m+n=2k_{1}+2k_{2}=2(k_{1}+k_{2}).$$

Como $k_{1}+k_{2}\in\mathbb{Z}$, a soma tem a forma $2k$ para algum inteiro $k$. Logo, $m+n$ é par.

**(II)** Sejam $m$ e $n$ dois inteiros ímpares. Existem $a,b\in\mathbb{Z}$ tais que $m=2k_{1}+1$ e $n=2k_{2}+1$. Desenvolvendo o produto, obtemos

$$
\begin{aligned}
mn&=(2k_{1}+1)(2k_{2}+1)\\
&=4k_{1}k_{2}+2k_{1}+2k_{2}+1\\
&=2(2k_{1}k_{2}+k_{1}+k_{2})+1.
\end{aligned}
$$

Como $2k_{1}k_{2}+k_{1}+k_{2}\in\mathbb{Z}$, o produto tem a forma $2k+1$ para algum inteiro $k$. Portanto, $mn$ é ímpar.
