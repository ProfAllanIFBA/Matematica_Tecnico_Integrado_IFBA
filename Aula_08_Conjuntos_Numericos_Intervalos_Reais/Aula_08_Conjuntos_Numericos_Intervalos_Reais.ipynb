## Introdução

Os números reais reúnem os números racionais e os irracionais. Nesta aula, vamos ver como esses números aparecem em medidas de comprimento e como podemos localizar alguns deles na reta real.

Nas aulas anteriores, vimos que $\sqrt{2}$ é irracional. Esse número aparece, por exemplo, quando medimos a diagonal de um quadrado de lado $1$.<br><br>

<center> <img src="fig1.jpg" alt="Logo do IFBA" width="100"> </center>

Pelo teorema de Pitágoras, o comprimento $d$ dessa diagonal satisfaz

$$
d^2=1^2+1^2=2.
$$

Como $d>0$, temos $d=\sqrt{2}$. Esse é o comprimento exato da diagonal. Em uma medição com régua, podemos obter um valor aproximado, como $1{,}41$, na mesma unidade de comprimento.

## 1.1 A reta real

###
<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Reta real</strong>: Em uma reta orientada, escolhemos uma origem, à qual associamos o número $0$, e uma unidade de comprimento. Assim, a cada número real corresponde um único ponto da reta, e a cada ponto corresponde um único número real.
</div>

Adotamos o sentido positivo para a direita. A uma unidade da origem, marcamos $1$ à direita e $-1$ à esquerda. Repetindo essa mesma unidade, podemos localizar $2$, $-2$ e os demais inteiros.

Se $a<b$, o ponto de coordenada $a$ está à esquerda do ponto de coordenada $b$.

Nos exemplos a seguir, usaremos $u$ para indicar a distância entre os pontos de coordenadas $0$ e $1$, tomando submúltiplos $\left(\frac{u}{2}, \frac{u}{3}, \frac{u}{4}, \ldots\right)$ ao dividir esse segmento em partes iguais e múltiplos $\left(2u, 3u, 4u, \ldots\right)$ ao colocar cópias dele uma após a outra.

#### Exemplo 1

Localize $1{,}5$ na reta real.

##### Solução

A partir da origem, a distância até o ponto de coordenada $1{,}5$ é $1{,}5u=u+\frac{u}{2}$. Portanto, marque o ponto $1{,}5$ meia unidade à direita de $1$.<br><br>

<center> <img src="fig2.jpg" width="250"> </center>

#### Exemplo 2

Localize $\dfrac54$ na reta real.

##### Solução

Divida o segmento de $0$ a $1$ em quatro partes iguais. Cada parte mede $\frac{u}{4}$. Como $\frac{5}{4}u=u+\frac{u}{4}$, marque o ponto de coordenada $\frac54$ um quarto de unidade à direita de $1$.<br><br>

<center> <img src="fig3.jpg" width="200"> </center>

#### Exemplo 3

Localize $\sqrt{2}$ na reta real.

##### Solução

1. Construa um quadrado usando como lado o segmento da reta entre $0$ e $1$. Esse lado mede $u$.
2. Trace a diagonal que parte da origem. Pelo teorema de Pitágoras, seu comprimento é $\sqrt{2}\,u$.
3. Coloque a ponta seca do compasso na origem e abra-o até o outro extremo da diagonal.
4. Com essa abertura, trace um arco que corte a reta à direita de $0$. O ponto de interseção tem coordenada $\sqrt{2}$.<br><br>

<center> <img src="fig4.jpg" width="250"> </center>

#### Exemplo 4

Localize $\sqrt{6}$ na reta real.

##### Solução

1. Use o **Exemplo 3** para construir o segmento que vai da origem ao ponto de coordenada $\sqrt{2}$. Esse segmento mede $\sqrt{2}\,u$.
2. Use esse segmento como base de um retângulo de altura $2u$.
3. Trace a diagonal que parte da origem. Pelo teorema de Pitágoras, seu comprimento $d$ satisfaz

   $$
   d^2=\left(\sqrt{2}\,u\right)^2+(2u)^2=6u^2.
   $$

   Como $d>0$, temos $d=\sqrt{6}\,u$.
4. Coloque a ponta seca do compasso na origem e ajuste a abertura até o outro extremo da diagonal. Trace um arco que corte a reta à direita de $0$. O ponto de interseção tem coordenada $\sqrt{6}$.<br><br>
